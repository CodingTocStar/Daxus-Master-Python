```python
# Funciones

# Ingresar producto
def ingresar_producto():
    nombre_producto = input("Nombre del producto: ").title().strip()

    existe = False
    for producto in inventario:
        if producto["Nombre"] == nombre_producto:
            existe = True
            break  # salgo del bucle for

    if existe == True:
        print(f"\n{nombre_producto} ya existe en el inventario.")

    else:
        cantidad_producto = int(input("Cantidad a ingresar: "))
        precio_unitario = float(input("Precio unitario: "))
        valor_stock = precio_unitario * cantidad_producto

        nuevo_producto = {
            "Nombre": nombre_producto,
            "Cantidad": cantidad_producto,
            "Precio unitario": precio_unitario,
            "Valor en stock": valor_stock
        }

        inventario.append(nuevo_producto)
        print()
        print(f"{nombre_producto} ingresado con exito.\n")

    print("-" * 40)


# Actualizar producto
def actualizar_producto():
    nombre_producto = input("Producto a actualizar: ").title().strip()

    producto_encontrado = None

    for producto in inventario:
        if producto["Nombre"] == nombre_producto:
            # asigo la variable momentanea del bucle a una que siga existiendo
            producto_encontrado = producto
            print(f"\nProducto encontrado.")
            break  # rompo el ciclo de busqueda

    if producto_encontrado is not None:
        nuevo_nombre_producto = input(
            f"Nuevo nombre de {nombre_producto}: ").title().strip()
        nueva_cantidad = int(input("Nueva cantidad: "))
        nuevo_precio_unitario = float(input("Nuevo precio unitario: "))
        nuevo_valor_stock = nueva_cantidad * nuevo_precio_unitario

        # Actualizo los valores
        producto_encontrado["Nombre"] = nuevo_nombre_producto
        producto_encontrado["Cantidad"] = nueva_cantidad
        producto_encontrado["Precio unitario"] = nuevo_precio_unitario
        producto_encontrado["Valor en stock"] = nuevo_valor_stock

        print(
            f"\n¡{nombre_producto} actualizado con éxito a '{nuevo_nombre_producto}'!\n")

    else:
        print(f"\n{nombre_producto} no existe en el inventario.\n")

    print("-" * 40)

# Eliminar producto
def eliminar_producto():
    producto_eliminar = input("Producto a eliminar: ").title().strip()
    producto_encontrado = None

    for producto in inventario:
        if producto["Nombre"] == producto_eliminar:
            producto_encontrado = producto
            break  # salgo del for

    if producto_encontrado is not None:
        inventario.remove(producto_encontrado)
        print(f"\n'{producto_eliminar}' eliminado exitosamente.\n")
    else:
        print(f"'\n{producto_eliminar}' no existe en el inventario.")

    print("-" * 40)

# Visualizar inventario
def abrir_inventario():

    for producto in inventario:
        print()
        print("*" * 40)
        print()
        print(f"Nombre: {producto["Nombre"]}")
        print(f"Cantidad: {producto["Cantidad"]}")
        print(f"Precio unitario: {producto["Precio unitario"]}")
        print(f"Valor en stock: {producto["Valor en stock"]}")

    print()
    print("*" * 40)

# Generar informe
def generar_informe():
    
    from fpdf import FPDF

    autor = input("Nombre de quien genera el informe: ")
    revisor = input("Nombre de quien revisa el informe: ")

    encabezado = ["Producto", "Cantidad", "Precio unitario", "Valor Total"]
    
    producto = [list(producto.values()) for producto in inventario]
    
    pdf = FPDF()
    pdf.add_page()
    pdf.set_font("helvetica", size=24, style="B") #B de negrita por bold
    
    # TITULO
    pdf.cell(w=190, h=10, text="Sistema de Gestion de Inventario", align="C") #añade una celda
    pdf.ln(20) #separacion de linea
    
    resumen = """
    En seguida se muestra el inventario actual para todos los productos despues de su gestion (añadir o eliminar productos)
    Tambien se muestra el total unificado del inventario en la tabla que s emuestra a continuacion
    """
    
    pdf.set_font("helvetica", size=16, style="B")

#Menu
def menu():
    print("Bienvenido al Sistema de Control de Inventario 5.0")
    opcion = ""

    while opcion != 0:
        print("\nElija una opcion:")
        print("1 - Ingresar nuevo producto.")
        print("2 - Actualizar producto.")
        print("3 - Eliminar producto.")
        print("4 - Abrir inventario.")
        print("5 - Exportar informe inventario actual en formato PDF.")
        print("0 - Finalizar programa.")

        opcion = int(input("Su eleccion: "))

        if opcion == 1:
            ingresar_producto()

        elif opcion == 2:
            actualizar_producto()

        elif opcion == 3:
            eliminar_producto()

        elif opcion == 4:
            abrir_inventario()

        elif opcion == 5:
            generar_informe()

        elif opcion == 0:
            print("\nFinalizando programa.\n")

        else:
            print("Ingrese una opcion valida!")
            print("-" * 40)

    print("=" * 40)
    print("\nPrograma finalizado.\n")

# Ejecucion del programa
inventario = [] #tiene que ser global o las funciones no van a poder acceder

menu()
```
