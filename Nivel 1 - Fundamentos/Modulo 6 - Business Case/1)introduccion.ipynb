# Business Case

* Es un documento o presentacion que justifica la realizacion de un proyecto, inversion o iniciativa dentro de una organizacion.
* Su proposito principal es *convencer* a los tomadores de decisiones de que vale la pena invertir recursos en la propuesta.

## Entonces:
* Resumen ejecutivo del caso
    * Es claro y breve del problema.
    * La solucion propuesta
    * Los beneficios esperados.
* Descripcion del problema
    * Situacion o necesidad a resolver.
* Plan de implementacion
    * Etapas del proyecto
    * Cronograma
    * Recursos necesarios

## OBjetivo de este business case

* Practicar todos los conecptos aprendidos a lo largo del curso
* Aplicar conceptos y teoria de los fundamentos a casos reales
* IMplementacion de **librerias** para problemas especificos.

## Escenario

* Contexto:
    * Sos parte de una organizacion que comercializa diferentes tipos de productos de consumo.
    * Se asigna a tu depto de logistica generar un sistema que permita a los clientes internos gestionar el inventario desde que se añaden productos (nombre, precio unitario, valor total del inventario), consulta inventario, hasta la generacion de un informe.

* <u>Requerimientos funcionales</u>: 
    * **Añadir** productos:
        * Nombre
        * Cantidad
        * precio unitario
        * Valor de stock
    * **Eliminar** productos:
        * Evitar ofrecer productos que ya no existen en la organizacion
    * **Consultar** estado actual del inventario:
        * Medir la capacidad de oferta
        * Puntos de abastecimiento
        * Etc
    * **EXPORTAR** informe en formato PDF

* <u>Requerimientos Tecnicos:</u>
    * Normalizacion del nombre del producto (title(), strip())
    * Separacion visual con caracteres ( "-" * 30) para un diseño mas organizado
    * Menu interactivo para que el usuario elija acciones (while())
    * Representacion del invetario de productos con estructuras de datos (list, dict)
    * Implementacion de la libreria fpdf. Para la exportacion del informe en formato PDF.
