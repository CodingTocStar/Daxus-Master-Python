# COnjuntos - `set()`

* son una estructura d datos que almacena elementos **<u>unicos</u>**

* Sin orden especifico.
* Definidos por {} o la funcion `set()`
* **<u>NO permiten elementos duplicados</u>**

## Principales caracteristicas

* Son **no ordenados** --> **NO** se acceden por <u>indice</u>
* **Mutables** --> *PERO...* no permiten <u>duplicados</u>
* **Iterables** --> Se pueden <u>recorrer</u> estos objetos

## Creando un `set`

```python
conjunto = {1,2,3,4,5}

print(conjunto)
print(type(conjunto))
```

```python
conjunto_repetido = {1,2,2,3,4,5}

print(conjunto_repetido) ## TOma solo el primero
```

## Manipulando los elementos de un `set`

### MEtodo `add()` --> SE pueden agregar de distintos tipos!

* Agrega el elemento x al conjunto
* Si ya existe --> NO hace nada (nada de duplicados)

```python
print(conjunto)
```

```python
conjunto.add(6)

print(conjunto)
```

```python
conjunto.add("Guts")
print(conjunto)
```

```python
conjunto.add(False)
conjunto.add(True) #LO toma como 1 y no lo agrega
print(conjunto)
```

```python
conjunto.add(1.234)
print(conjunto)
```

```python
conjunto.add(("Naruto", "Uzumaki")) ## Podes meter tuplas

print(conjunto)
```

### NO permiten listas, las listas permiten duplicados

```python
# conjunto.add(["1,2,3"]) || TypeError unhasheable, ya estan presentes
# conjunto.add(["7,8,9"])
print(conjunto)
```

## Eliminando elementos de un `set`

### Metodo `remove()`

* Elimina el elemento x del conjunto
* Si el elemenot no esta presente --> KeyError 

```python
mi_conjunto = {1,2,3,4}
print(mi_conjunto)

mi_conjunto.remove(3)
print(mi_conjunto)
```

## Metodo `pop()`

* Elimina y **devuelve** un elemento <u>aleatorio</u> del conjunto.
* Si el conjunto esta vacio --> KeyError

```python
mi_conjunto = {1,2,3,4}
print(mi_conjunto)

elemento = mi_conjunto.pop()

print(elemento)

print(mi_conjunto)
```

## Metodo `clear()`

Elimina todos los elementos del conjunto

```python
print(mi_conjunto)
mi_conjunto.clear()

print(mi_conjunto)
```

# Operaciones con `sets`


## Metodo `union()`

Devuelve un conjunto con los elemenos combinados de dos conjuntos (**sin duplicados**)

```python
conjunto1= {1,2,3}
conjunto2 = {4,2,5}

print(conjunto1.union(conjunto2))
print(conjunto1 | conjunto2)
```

## Metodo `interseccion()`

Devuelve un conjunto con los elementos **comunes** entre dos conjuntos.

```python
conjunto1 = {1,2,3}
conjunto2 = {2,3,5}

print(conjunto1.intersection(conjunto2))
print(conjunto1 & conjunto2)
```
