# Sistema control de inventario

## Funcionalidades

* `listas` --> para almacenar multiples productos sin usar estructuras de repeticion o control
* `diccionarios` --> para estructurar los datos (`nombre`, `cantidad`, `precio _unitario`, `valor_total`)
* Manipulacion de cadenas (`strip()`, `title()` para estandarizar la entrada del usuario)
* Calculo automatico del valor total del inventario
* Separacion visual con caracteres (`"-" * 48`)
