# Sistema control de inventario

## Funcionalidades

* `listas` --> para almacenar multiples productos sin usar estructuras de repeticion o control
* `diccionarios` --> para estructurar los datos (`nombre`, `cantidad`, `precio _unitario`, `valor_total`)
* Manipulacion de cadenas (`strip()`, `title()` para estandarizar la entrada del usuario)
* Calculo automatico del valor total del inventario
* Separacion visual con caracteres (`"-" * 48`)

```python
# Creando inventario
inventario = []

```

```python
# Inicio programa
print("Bienvenido al Sistema de Control de Inventario 3.0")
nombre_producto = input("Ingrese el nombre del producto: ").title().strip()

cantidad_producto = int(input("Ingrese cantidad: "))

precio_unitario = float(input("Ingresa el precio unitario: "))

precio_total = precio_unitario * cantidad_producto

producto1 = {
    "Nombre": nombre_producto,
    "Cantidad": cantidad_producto,
    "Precio_unitario": precio_unitario,
    "Precio_total": precio_total
}

inventario.append(producto1)

nombre_producto = input("Ingrese el nombre del producto 2: ").title().strip()

cantidad_producto = int(input("Ingrese cantidad: "))

precio_unitario = float(input("Ingresa el precio unitario: "))

precio_total = precio_unitario * cantidad_producto

producto2 = {
    "Nombre": nombre_producto,
    "Cantidad": cantidad_producto,
    "Precio_unitario": precio_unitario,
    "Precio_total": precio_total,
}


inventario.append(producto2)

print("*" * 40)
print("Sus productos:")
print("Nombre      Cantidad      Precio unitario      Precio Total")
print(
    f'{inventario[0]["Nombre"]}      {inventario[0]["Cantidad"]}     {inventario[0]["Precio_unitario"]}     {inventario[0]["Precio_total"]}'
)
print(
    f'{inventario[1]["Nombre"]}      {inventario[1]["Cantidad"]}     {inventario[1]["Precio_unitario"]}     {inventario[1]["Precio_total"]}'
)
```
