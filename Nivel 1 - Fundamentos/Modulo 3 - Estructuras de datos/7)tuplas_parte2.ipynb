## Desempaquetado de una tupla

* Extrae los elementos de una tupla y los asigna a variables de manera simultanea.
* Forma concisa y eficiente de acceder a los elementos de una tupla (o cualquier iterable) y asignarlos a variables individuales en una sola linea

## Desempaquetado basico

```python
simpsons = ("Homero", "Marge", "Lisa", "Bart")

padre, madre, hija, hijo = simpsons

print(padre)
print(madre)
print(hijo)
print(hija)
```

## Desempaquetando con tuplas de longitud variable

```python
print(simpsons)
padre, *resto_familia = simpsons

print(padre)
print(type(padre))
print()
print(resto_familia)
print(type(resto_familia))
```

## Desempaquetado ignorando elementos

* Si solo te interesa algunos de los elementos de la tupla
    * Podes ignorar los elementos que **NO** te interesan usando el simbolo de subrayado(_) --> convencion comun



```python
tupla = (1,2,3,4)

a, _, c, _ = tupla

print(a)
print(c)
print("-----------")
print(_) #se reemplaza el primer _ y queda el ultimo
```
