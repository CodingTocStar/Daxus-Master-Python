## Operadores in y not in

```python
personajes = ["Naruto", "Luffy", "Pikachu", "Bart", "Homero", "Usopp"]

print("Gohan" in personajes)
print("Bart" in personajes)

print("Homero" not in personajes)

```

## Manipulando elementos de una lista

## Metodo append()

Añade un elemento al final de la lista

```python
print(personajes)
personajes.append("Maggie")
print(personajes)
```

## Metodo insert()

Lo mete en una posicion especifica de la lista

```python
print(f"Personajes antes de aplicar el metodo insert: {personajes}")

```

## Metodo count()

Cuenta cuantas veces aparece un valor en la lista

```python
lista_numeros = [12,123, 15452, 1256, 565, 12, 123, 123]

print(lista_numeros)
```

```python
print(lista_numeros.count(123))
```

## Metodo reverse()
INvierte el orden de los elementos de la lista

```python
mi_lista = [1,2,3]

mi_lista.reverse()

print(mi_lista)
```

## Metodo sort()
* Ordena la lista en orden ascendente.
* Con reverse = True, de forma descendente

```python
mi_lista = [3,1,2]

mi_lista.sort()

print(mi_lista)

mi_lista.sort(reverse=True)

print(mi_lista)
```

## Metodo pop()

* Elimina y devuelve le elemento eb yba oisucuib especifica
    * Por defectoo elimina el ultimo

```python
personajes = ["Naruto", "Luffy", "Pikachu", "Bart", "Homero", "Usopp"]


print(personajes)

print(personajes.pop())

print(personajes)
```

```python
personajes = ["Naruto", "Luffy", "Pikachu", "Bart", "Homero", "Usopp"]


print(personajes)

print(personajes.pop(-2))

print(personajes)
```

## Metodo remove()

elimina la primera aparicion de un valor de la lista

```python
lista_repetidos = [1,2,3,4,4,5]
print(lista_repetidos)
```

```python
lista_repetidos.remove(4)

print(lista_repetidos)
```

## Metodo clear()

* Elimina todos los elementos de la lista
    * La deja vacia

```python
print(lista_repetidos)
```

```python
lista_repetidos.clear()
print(lista_repetidos)
```

## Comando del

Podemos exluir un elmento de una lista utilizando el comando del, seguido de la lista y el indice del elemento que sera excluido

```python
lista_lenguajes = ["Python", "Java", "C++"]

print(lista_lenguajes)
```

```python
del lista_lenguajes[1]

print(lista_lenguajes)
```
