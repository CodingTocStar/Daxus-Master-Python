# *args - Parametros posicionales arbitrarios

* Permite que una funcion reciba un numero variable de argumentos posicionales (o sea sin nombrar)
* Todos los argumentos que se oasen a la funcion ( ademas de los parametros definidos antes de *args) se agruparan en una **tupla**

```python
def suma_dos(num1, num2):
    respuesta = num1 + num2
    return respuesta


print(suma_dos(5,3))
print(suma_dos(5,3,3))
    
```

```python
def suma_usuario(nombre, *args):
    print(f"{nombre.title()}, la suma de los numeros {args} es: {sum(args)}")
    
#invocar
suma_usuario("Cete", 1,2,3,4)

suma_usuario("Felipe", 123, 342, 543, 432432, 424)
```

## *kwargs - Parametros posicionales arbitrarios

* Similar al anterios
* Recibe argumentos con nombre (pares *clave-valor*)
* Se agrupan en un *diccionario*
    * Las claves son los nombres de los parametros.
    * Los valores son los valores pasados.



```python
def mostrar_datos_usuario(**kwargs):
    for k, v in kwargs.items():
        print(f"{k.title()} : {v.title()}")
```

```python
mostrar_datos_usuario(nombre = "Naruto",  apellido = "uzumaki", aldea = "konoha")
```

# El orden en que se pasan los argumentos es **CLAVE**

1. Parametros *posicionales* -> obligatorio
2. Parametros con *valores por defecto*
3. **args* -> cualquier numero adicional de parametros posicionales
4. Parametros *nombrados* -> parametro obligatorio que solo se puede pasar por nombre (keyword)
5. ***kwargs*

## Ejemplo con todos los tipos de parametros

```python
def mi_funcion(posicional, valor_por_defecto = 2, *args, nombrado, por_defecto = 4, **kwargs):
    print(posicional, valor_por_defecto, args, nombrado, por_defecto, kwargs)
```

```python
mi_funcion(1,3,5,6, nombrado = 7, nombre = "Juancho")
```
