# docstring

* Cadena de texto(str) que se utilizan para *documentar* el proposito y funcionamiento de una funcion.
* Estan justo despues de la definicion de la funcioin, clase o modulo.
* Sirven como documentacion interna
    * Describe:
        * Que hace el codigo
        * Como se usa
        * Parametros esperados
        * VAlores devueltos

## Caracteristicas

* Se colocann entre comillas triples (""" o ''')
* Para acceder en tiempo de ejecucion se usa el atributo especial `__doc__`

```python
personajes = []

def agregar_personajes(nombre_usuario):
    """_sumary_
    
    Esta funcion agrega a una lista el personaje suminstrado por el usuario
    
    Args: 
        nombre_usuario (str): Nombre del usuario para personalizar
    
    Returns:
        None.
    """
    
    while True:
        personaje = input(f"{nombre_usuario.title()}, ingresa el personaje que quieres agregar: ").title()
        
        if personaje.isdigit() or personaje.isnumeric() or personaje.isdecimal():
            print(f"Valor no valido {personaje}. {nombre_usuario.title()}, intenta de nuevo...")
        
        elif(not personaje.isdigit() or not personaje.isnumeric() or not personaje.isdecimal() and personaje != "Salir" and personaje != ""):
            print(f"{nombre_usuario.title()}, has agregado al personaje '{personaje}'")
            personajes.append(personaje)
    
            print(f"La lista de personajes: {', '.join(personajes)}\n")
            
        elif personaje == "Salir":
            print(f"Saliedno del programa..-. 🤖 Chau {nombre_usuario}")
            print(f"Lista final de personajes: {', '.join(personajes)}")
            break
    
```

```python
nombre_usuario = input("Ingresa tu nombre: ").title()

agregar_personajes(nombre_usuario)
```

```python
print(agregar_personajes.__doc__)
```
