# TIpos de parametros

* posicionales
* nombrados

## Parametros posicionales

Se pasan a la funcion en el mismo orden en que fueron declarados.

```python
def saludo_usuario(nombre, apellido):
    print(f"{nombre.title()} {apellido.title()} bienvenido al curso de python")
    print(f"{nombre.title()} ya estas registrado en eln sistema")

saludo_usuario("Naruto", "uzumaki")
```

## Parametros con valor predeterminado

Si no se pasan al llamar la funcion se usa un valor predeterminado

```python
def registro_empleado(edad, nombre, estado_civil = "Soltero", compania = "Python Corp.", status = "activo"):
    print(f"El empleado {nombre} tiene la {edad} años de edad, y su estado civil es {estado_civil}. Ingresa a la compañia {compania} y su estado es {status}")
    
```

```python
#Invoco la funccion

registro_empleado(32, "Vegeta")

registro_empleado(32, "Vegeta", estado_civil= "Casado")

registro_empleado(32, "Vegeta", estado_civil="Casado", status= "Capacitacion")
```

## Parametros nombrados

Se pasan explicitamente por su nombre al llamar a la funcion, sin importar el orden.

```python
def saludo_usuario(nombre, apellido):
    print(f"{nombre.title()} {apellido.title()} bienvenido al curso de python")
    print(f"{nombre.title()} ya estas registrado en eln sistema")
    
saludo_usuario(apellido="uzumaki", nombre="naruto")
```
