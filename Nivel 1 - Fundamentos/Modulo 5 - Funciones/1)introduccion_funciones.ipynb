# Tipos de funciones

## QUe son?

Son bloques de codigo reutilizable

* Built-in -> son parte de python, las usas y ya
* custom -> definidas por el usuario
    * pueden retornar o no un valor

## Porque usarlas?

* Reutilizacion del codigo
* Modularidad y organizacion
    * Division del programa en partes pequeñas
    * cada funcion ejecuta una tarea mas pequeña
* Depuracion y mantenimieto
    * Facilidad de identificacion de problemas
    * Si necesitas cambiar una funcionalidad solo modificas la funcion
* Abstraccion
    * Se puede ocultar la complejidad de alguna tarea
    * Proporciona una interfaz mas sencilla
* Colaboracion
    * Cada funcion puede estar a cargo de un integrante del grupo
* Facilita la lectura y comprension
    * Mas sencillo de leer con funciones que describen la tarea
