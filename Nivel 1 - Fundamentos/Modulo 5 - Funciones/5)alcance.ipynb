# ALcance de las funciones (function scope)

* Local 
* Global
* Scope de las funcioens anidadas (*enclosing scope*)

## Local Scope

* Son las que estan definidas <u>dentro de la funcion</u>
* Solo son accesibles dentro de la misma

```python
def funcion():
    x = 42
    print(f"Valor de la variable definiido dentro de la funcion: {x}")
    
funcion()
print(x)
```

```python
def validar_palabra():
    palabra = input("ingresa una palabra: ").title()
    
    if palabra == "Python":
        print(f" Excelente, escribiste '{palabra}, la plabra que representa al lenguaje mas utilizado")
    else:
        print(f"Escribiste '{palabra}', no es la que esperabamos")

validar_palabra() 

validar_palabra()  

print(palabra)     
```

## Global scope

* Definidas *fuera* de cualquier funcion.
* Accesibles en cualquier parte del codigo
    * Incluso *dentro* de las funciones.
    * Salvo que haya una variable local con el mismo nombre



```python
variable_global = 42

def mi_funcion():
    print(f"Accediendo desde la funcion 'mi_funcion' a la variable global: {variable_global}")
    
def otra_funcion():
    print(f"Accediendo desde la funcion 'otra_funcion' a ka variable global: {variable_global}")
    
mi_funcion()
otra_funcion()
print(variable_global)
```

## Enclosing scope

* SI una funcion esta definida dentro de otra, las varialbes de la funcion exterior son accesibles desde la interior
* **NO** al reves

```python
def funcion_externa():
    variable_externa = 10
    print(f"LLamada desde la funcion externa")
    print(f"Variabke extenra: {variable_externa}")
    print(var_interna)
    print()
    
    def funcion_interna():
        var_interna = 4
        print(f"Llamada desde la funcion interna")
        print(f"Variable interna_ {var_interna}")
        print(f"Varialbe externa: {variable_externa}")
    
    funcion_interna()
```

```python
funcion_externa()
```

`global`

* PErmite modificar una variable global dentro de una funcion

```python
variable_global = 5

def modificar_global():
    global variable_global
    variable_global = 10

modificar_global()
print(variable_global)
```

## `nonlocal`

* Cuando queres modificar una variable de una funciion exterior (pero no global) desde una funcion interna.

```python
def funcion_exterior():
    var_externa = 20

    print(f"LLamada desde la funcion exterior")
    print(f"Variabke exterior: {var_externa}")
    
    def funcion_interna():
        nonlocal var_externa
        var_externa = 25
        print(f"Llamada desde la funcion exterior")
        print(f"Variable exterior: {var_externa}")
    
    funcion_interna()
```

```python
funcion_exterior()
```
