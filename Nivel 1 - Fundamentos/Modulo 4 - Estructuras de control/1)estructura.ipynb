# Estructura de repeticion

Permiten ejecutrar una instruccion o un bloque de codigo varias veces.

* usan un contador o condicion para la ejecucion
* for: utera sobre una secuencia (lista, tupla, rango, etc)
    * ejecuta un bloque de codigo para cada elemento
* while: ejecuta un bloque de codigo mientras se cumpla una condicion
