# sistema control de inventario

* Menu interactivo para que el usuario elija opciones
* registro de multiples productos sin limite predefinido
* eliminacion de prodfuctosd del inventartio
* visualizacion  del inventario actualizado
* uso de while for para recorrer los productos
* uso de condicionales para gestionar el menu

```python
print("Bienvenido al Sistema de Control de Inventario 4.0")
opcion = ""
inventario = {}


while opcion != 0:
    print("que desea hacer?")
    print("1 - Ingresar nuevo producto.")
    print("2 - Actualizar producto")
    print("3 - Eliminar producto")
    print("4 - Visualizar inventario")
    print("0 - Finalizar programa")

    opcion = int(input("Ingrese su opcion"))

    if opcion == 1:
        nombre_producto = input("Ingrese nombre del producto: ").title().strip()
        cantidad_producto = int(input("Ingrese cantidad: "))
        precio_unitario = float(input("Ingrese el precio unitario: "))
        precio_total = precio_unitario * cantidad_producto

        producto = {
            nombre_producto: {
                "Cantidad": cantidad_producto,
                "Precio unitario": precio_unitario,
                "Precio total": precio_total    
            }    
        }
        
        if nombre_producto not in inventario.keys():
            inventario[nombre_producto] = producto[nombre_producto] # en el inventario en la llave nombre_producto agrego lo que me da al acceder al dict producto en su llave nombre_producto
            print(f"Producto {nombre_producto} ingresado correctamente.")
        else:
            print(f"El producto {nombre_producto} ya existe en el inventario.")
        #inventario.update(producto)
        
        print("-" * 30)
    elif opcion == 2:
        nombre_producto = input("Que producto quiere actualizar?").title().strip()
        nuevo_nombre_producto = input("Que nombre nuevo le pone?").title().strip()
        
        cantidad_producto = int(input("Ingrese la nueva cantidad: "))
        precio_unitario = float(input("Ingrese el nuevo precio unitario: "))
        precio_total = precio_unitario * cantidad_producto

        producto = {
            nuevo_nombre_producto: {
                "Cantidad": cantidad_producto,
                "Precio unitario": precio_unitario,
                "Precio total": precio_total    
            }    
        }
        
        inventario[nuevo_nombre_producto] = inventario[nombre_producto]
        del inventario[nombre_producto]
        inventario[nuevo_nombre_producto] = producto[nuevo_nombre_producto]
        
        print(f"Producto {nombre_producto} actualizado correctamente a {nuevo_nombre_producto}.")
        print("-" * 30)
    elif opcion == 3:
        print()
        producto_a_eliminar = input("Que producto quiere eliminar?").title().strip()
        inventario.pop(producto_a_eliminar)
        print(f"Ha eliminado {producto_a_eliminar} exitosamente.")
        print("-" * 30)
    elif opcion == 4:
        print("")
        print("")
        for nombre, datos in inventario.items():
            print(
                f'Producto: {nombre}    Cantidad: {datos["Cantidad"]}    Precio unitario: {datos["Precio unitario"]}    Precio total: {datos["Precio total"]}')
        print("-" * 30)
    elif opcion == 0:
        print("NOs vimos!")
    else:
        print("Ingrese una opcion valida")
        print("-" * 30)

print("Programa finalizado.")
```
