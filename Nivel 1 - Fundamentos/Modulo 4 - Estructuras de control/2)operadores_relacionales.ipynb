# Operadores relacionales

* mayor >
* mayor igual >=
* Menor <
* Menor igual <=
* igual == (comparacion)
* diferente !=

```python
edad_padre = 50
edad_hija = 12

print(edad_padre < edad_hija)
```

```python
distancia1 = 3.5
distancia2 = 8

print(distancia2 >= distancia1)
```

```python
ciudad_usuario1 = "Bogota"
ciudad_usuario2 = "Cali"

print(ciudad_usuario1 == ciudad_usuario2)
```

```python
temperatura1 = 18
temperatura2 = 18

print(temperatura1 != temperatura2)
```
