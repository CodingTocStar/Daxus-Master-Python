# Que estructura de control de repeticion usar?

## Cuando for

* Cuando el numero de iteraciones es conocido
    * rango
    * objeto
    * iterable
* Tenes una estructura de datos a procesar


## cuando while

* Si el numero de iteraciones **no es conocido**
* Dependes de una condicion que cambia en la ejecucion del ciclo
* La condicion es dinamica o depende de algun evento o de entrada externa.
* El ciclo debe continuar hasta que la condicion cambie de manera **impredecible**
* **Control** mas detallado sobre la condicion que termina el ciclo (break)
