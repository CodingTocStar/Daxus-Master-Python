# Estructuras de control

## Condicional if

Ejecuta bloques de codigo si la condicion es verdadera o falsa

if "condicion":

    bloque que se ejecuta si la condicion es verdadera

Su sintaxis compuesta es la siguiente:

if "condicion":
    bloque ejecutado si es verdadera
else:
    ejecutado si es falsa

* Se pueden usar todos los operadores relacionales y logicos

## SIntaxis basica if


```python
edad = int(input("Ingresa tu edad:"))

if edad >= 18:
    print(f"La personma tiene {edad} años de edad")
    print("La persona es mayor de edad")
```

## Sintaxis compuesta if con else

```python
edad = int(input("Ingrese tu edad:"))
licencia_conducir = True

if edad >= 18 and licencia_conducir == True:
    print(f"La persona tiene {edad} años de edad")
    print("La persona es mayor de edad")
    print("Esta habilitada para conducir")
    print(f"Habilitado para conducir: {True}")

else:
    licencia_conducir = False
    print(f"La persona tiene {edad} años de edad")
    print("La persona es no mayor de edad")
    print(f"Habilitado para conducir: {licencia_conducir}")
```

## SIntaxis compuesta if-elif-else

```python
edad = int(input("Ingrese tu edad:"))
licencia_conducir = True

if edad >= 18 and licencia_conducir == True:
    print(f"La persona tiene {edad} años de edad")
    print("La persona es mayor de edad")
    print(f"Habilitado para conducir: {True}")

elif edad >= 16 and edad < 18 and licencia_conducir:
    print(f"La persona tiene {edad} años de edad")
    print("La persona aun no es mayor de edad")
    print(f"Habilitado para conducir: {licencia_conducir}")

else:
    licencia_conducir = False
    print(f"La persona tiene {edad} años de edad")
    print("La persona es no mayor de edad")
    print(f"Habilitado para conducir: {licencia_conducir}")
```

## if ternario

* Es una forma compacta de escribir la estrucuta en una sola linea
* Sirve cuando queres asignar un valor a una variable dependiendo de una condicion

### Sintaxis

"sentencia true" if condicion else "sentencia false"

```python
edad = int(input("Ingrese tu edad:"))
licencia_conducir = True

mensaje = "Eres mayor de edad" if edad>= 18 else "Eres menor de edad"

print(mensaje)
```
