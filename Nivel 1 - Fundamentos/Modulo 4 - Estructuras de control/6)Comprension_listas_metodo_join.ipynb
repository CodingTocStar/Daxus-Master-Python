# list comprehension y el metodo join()

## `list comprehension`

* Es una forma concisa y elegante de crear listas.
* Permite crear una nueva lista aplicando una expresion a cada elemento de una secuencia o iterable en **una linea de codigo** 
    * lista
    * rango
    * etc

<u>>Sintaxis basica</u>

[nueva_expresion] for elemento in iterable

<u>Sintaxis con condicional</u>

[nueva_expresion] for elemento in iterable if condicion

```python
lista_numeros = [i for i in range(1,11)]

print(lista_numeros)
```

```python
numeros_cuadrados = [elemento**2 for elemento in lista_numeros]

print(numeros_cuadrados)
```

```python
numeros_pares = [elemento for elemento in lista_numeros if elemento%2 == 0]

print(numeros_pares)
```

## MEtodo join()

* Se usa para unir secuencias de elementos (generalmente una lista de cadenas de texto) en una sola cadena, con un **delimitador** especifico entre esos elementos.

<u>Sintaxis</u>

delimitador.join(iterable)


```python
personajes = ["goku", "tanjiro", "luffy", "naruto"]

```

```python
print(f"Mis personajes favoritos son: {" - ".join(personajes)}")
```

```python
personajes = [personaje.title() for personaje in personajes]

print(f"Mis personajes favoritos son: {", ".join(personajes)}")
```
