# Esctructura de repeticion for

* Se usa para iterar (recorrer) sobre una secuencia, como lista, tupla, diccionario, conjunto o incluso una cadena de texto.
* Sirve cuando sabes cuantas veces queres repetir un bloque de codigo.

Su sintaxis es la siguiente: 

for "variable" in "secuencia":
    Bloque a ejecutar por cada elemento de la secuencia

```python
vehiculos = ["moto", "carro", "bici", "avion", "helicoptero", 3, False]

print(vehiculos[0])
print(vehiculos[1])
```

```python
for i in vehiculos:
    print(i)
```

```python
numeros = [1,2,3,4,5]

for numero in numeros:
    print(f"Numero original")
```
