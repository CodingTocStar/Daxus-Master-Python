# Slicing (segmentacion)

Es una tecnica que extrae una porcion o subsecuencia de una cadena de texto.

## Accediendo a un indice

```python
frase ="Mi nombre es naruto y estoy aprendiendo python"

print(frase[0])
print(frase[1])
print(frase[11])
```

## Acceder a un substring

## sintaxis

string[start:stop:step]

```python
print(frase)
```

```python
print(frase[0:9])
```

```python
print(frase[0:19])
```

```python
print(frase[-6:])
```

Con el paso

```python
print(frase[0::2])
```

## invirtiendo un str

```python
frase = "anita lava la tina"
```

```python
print(frase[::-1])
```

```python
frase ="Mi nombre es naruto y estoy aprendiendo python"
print(frase[::-1])
```
