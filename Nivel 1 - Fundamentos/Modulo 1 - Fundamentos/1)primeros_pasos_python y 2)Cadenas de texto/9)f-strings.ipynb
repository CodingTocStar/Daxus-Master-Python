# f-strings

* string literal formateada (formated string literl)
    *  cadenas sencilla y eficientemente


```python
nombre = "miguel"
edad = 17

print(nombre)
print(f"Hola mi nombre es {nombre} y tengo {edad} años")
```

```python
num1 = int(input("ingrese el primer numero:"))
num2 = int(input("ingrese el segundo numero:"))

print(f"La suma de los numeros {num1} y {num2} es igual a {num1 + num2}")
```

## Otros metodos


### Metodo format

```python
nombre = "vegeta"
edad = 27

print("hola soy {} y tengo {} años." .format(nombre, edad))
```

### Metodo %

```python
print("Hola miu  es %s y tengo %d años." %(nombre, edad))
```
