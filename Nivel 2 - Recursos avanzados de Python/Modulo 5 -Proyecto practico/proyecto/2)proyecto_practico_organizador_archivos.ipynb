```python
lista_archivo = [archivo for archivo in os.listdir() if os.path.isfile(archivo) and archivo.split(".")[-1] and archivo.split(".")[-1] != "ipynb" and archivo.split(".")[-1] != "py"]

lista_archivo
```

## 2. Identificar los diferentes `tipos` de archivos

```python
#lista_tipos = [tipo.split(".")[-1] for tipo in lista_archivo]
lista_tipos = set([tipo.split(".")[-1] for tipo in lista_archivo])

# SET COMPREHENTION
tipos_unicos = {tipo.split(".")[-1] for tipo in lista_archivo if tipo.split(".")[-1] != "ipynb" and tipo.split(".")[-1] != "py"}

tipos_unicos
```

## 3. Crear las carpetas para cada tipo de archivo

```python
for tipo in lista_tipos:
    if not os.path.isdir(tipo):
        os.mkdir(tipo)
        print(f"La carpeta '{tipo}' fue creada con exito.")
    else:
        print(f"La carpeta '{tipo}' ya existe.")
        
```

## 4. Mover los archivos a sus correspondientes carpetas

```python
for archivo in lista_archivo:
    carpeta_destino = archivo.split(".")[-1]
    origen = os.path.join(os.getcwd(), archivo)
    destino = os.path.join(os.getcwd(), carpeta_destino, archivo)
    print(f"Archivo '{archivo}' -> Carpeta '{carpeta_destino}'")
    os.rename(origen, destino)
```
