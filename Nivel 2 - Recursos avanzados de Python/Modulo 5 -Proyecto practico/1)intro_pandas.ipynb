# Introduccion

* pandas es una libreria diseñada para tareas de analisis y manipulacion de datos
* Orientado al manejo de datos de caracter tabular, como:
    * las hojas de calculo 
    * Bases de datos relacionales
* Su funcionalidad gira en torno a dos estructuras de datos fundamentales:
    * `Series`: 
        * Arreglo unidimensional de datos, cuyos elementos son identificados por un indice (no necesariamente unico como de los diccionarios)
        * Se pueden considerar como dos arreglos de NumPy asociados:
            * El indice de la serie -> sus valores corresponden a los valores usados para identificar cada entrada de la serie.
            * El contenido de la serie --> del mismo tamaño que su indice.
    * `Dataframes`:
        * `pd.DataFrame`, El objeto principal de pandas.
        * Representan una tabla, un tipo de conjutno de datos con un indice asignado a sus filas y uno a sus columnas.
        * Cada fila o columna representa un objeto `pd.Series`

## MEtodo `read_csv()`

* `pandas` permite cargar archivos en forma de `DataFrame`.
* Los argumentos de la funcion son **fundamentales**. 
    * Posibles problemas al cargar inadecuadamente un archivo.
* Algunos argumentos de los mas importantes a la hora de la carga de archivos en formato `.csv` son:
    * `path`: Ruta del archivo a cargar.
        * Es el primero en posicion y es **OBLIGATORIO**.
        * Archivos locales y URLs remotas.
    * `sep`: Separador de los elementos del archivo.
    * `header`: Posicion del header.
        * La fila indicada se interpreta como los nombres de las columnas y se cargaran las filas que le precedan.
    * `names`: Nombres a usar para las columnas cargadas.
        * La longitud de esta lista/arreglo debe **coincidir** con el **numero de columnas**
    * `index_col`: Posiciones de las etiquetas usadas como indice.
        * Si es una lista se crea un indice multiple.
        * Si se pasa el valor `False` se puede indicar a pandas que no use la primera columna como el indice del *DataFrame*
    * `usecols`: Subconjunto de las columnas cargadas a usar en la generacion del dataframe
    * `true_values` | `false_values`: Caracteres que seran interpretados como valores de verdad.
    * `skip_rows` | `skip_footer`: FIlas a saltar desde el inicio y desde el final.
    * `nrows`: Total de filas a cargar.
        * Especialmente util para conjuntos de datos grandes donde no nos interese toda la informacion.
    * `na_values` | `keep_default_na` | `na_filter`: Herramientas para el tratamiento d datos faltantes.
        * Muy importante en el proceso de limpieza de datos.
* Ademas de estos existen otros argumentos para la definicion del formato de origen como:
    * fecha
    * numeros decimales
* Esto ofrece variedad de opciones en la lectura de conjuntos de datos.
