# TRabajando con archivos csv

* csv: Comma Separeted Values. Pueden usar como separador ";", "\t"
* es un archivo de texto
* El modulo csv nos permite trabajar con estos archivos
    * Retorna un objeto **_csv.reader**

## Leyendo un archivo csv

```python
with open("msft.csv", mode="r") as f:
    data = f.read()

print(data)
```

```python
import csv
```

```python
with open("msft.csv", mode="r") as f:
    data_csv = csv.reader(f, delimiter=",")

    for dato in data_csv:
        print(dato)

print(type(data_csv))
```

## Creando un archivo csv

```python
import csv

datos = [
    ["nombre", "Edad", "Ciudad"],
    ["Juan", "25", "Madrid"],
    ["Ana", "22", "barcelona"],
    ["Jhoan", "35", "Valencia"],
]

with open("archivo_personas.csv", mode="w", newline="\n") as archivo_csv:
    escritor_csv = csv.writer(archivo_csv)
    
    escritor_csv.writerows(datos)

print("Archivo csv fue creado con exito")
```

# Trabajando con archivos JSON

* JSON: Javascript Object Notation
* Es un Archivo de texto
* Formato de archivo compacto y facil de leer para las personas
* Comunmente utilizado para:
    * intercamvio ede datos en la web
    * almacenamiento en bases de datos no relacionales (NoSQL)
* El modulo `json` nos permite trabajar con este tipo de archivos.
