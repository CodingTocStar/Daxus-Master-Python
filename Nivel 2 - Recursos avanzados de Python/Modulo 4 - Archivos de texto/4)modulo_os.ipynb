# Manejo de sistema de archivos con `os`

Ventajas de python que es de alto nivel es que: 

* nos da una interfaz unificada oara la manipulacion del sistema de archivos desde cualquier OS

## Metodo `getcwd()`
get current working directory

Revisar el `directorio` actual

```python
import os 
```

```python
print(os.getcwd())
```

## Metodo `listdir()`

```python
print(os.listdir("."))
print(os.listdir("./modulo_os/"))
print(os.listdir("./modulo_os/carpeta/"))
```

## Metodo `join()` y `basename()`

```python
os.path.join(os.getcwd(), "nuevo_archivo.txt")
```

```python
os.path.basename(os.getcwd())
```
