## Metodo `getcwd()`
get current working directory

Revisar el `directorio` actual

```python
import os 
```

```python
print(os.getcwd())
```

## Metodo `listdir()`

```python
print(os.listdir("."))
print(os.listdir("./modulo_os/"))
print(os.listdir("./modulo_os/carpeta/"))
```

## Metodo `join()` y `basename()`

```python
os.path.join(os.getcwd(), "nuevo_archivo.txt")
```

```python
os.path.basename(os.getcwd())
```

## Metodo `isfile()` y `isdir()`

```python
print(os.path.isfile("biblia.txt"))
print(os.path.isfile(os.path.join(os.getcwd(), "modulo_os/carpeta/archivo.txt")))
print(os.path.isfile("modulo_os"))



```

```python
print(os.path.isdir("modulo_os"))
print(os.path.isdir("biblia.txt"))
```

## Metodos `mkdir()` y `rmdir()`

make dir y remove dir

```python
os.mkdir("carpeta2")
```

```python
os.listdir()
```

```python
os.rmdir("carpeta2")
print(os.rmdir("carpeta2"))
print(os.listdir())
```

## Metodos `rename()` y `replace()`

rename sirve para:
* renombrear
* mover archivos entre directorios

```python
# Para renombrar archivos y/o carpetas con rename

#os.rename("carpeta/archivo.txt", "carpeta/archivo1.txt")
print(os.listdir(os.path.join(os.getcwd(), "carpeta")))
```

```python
os.replace("carpeta", "carpeta10")
```

```python
# Para mover archivos y/o carpetas con rename

origen = os.path.join(os.getcwd(), "carpeta10/archivo1.txt")
destino = os.path.join(os.getcwd(), "archivo1.txt")
```

```python
os.rename(origen, destino)
```

## Metodo `remove()`

```python
os.remove(os.path.join(os.getcwd(), "archivo1.txt"))
```

```python
os.listdir(os.getcwd())
```
