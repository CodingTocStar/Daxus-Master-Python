## Escribiendo una frase conla funcion open() y write()

```python
archivo = open("mi_archivo.txt", mode="w", encoding="utf-8")
```

```python
frase = "Mi nombre es Luffy y estoy aprendiendo python "
archivo.write(frase)
archivo.close()
```

## Escribiendo varias frases con la funcion open() y write()

```python
archivo = open("mi_archivo.txt", mode="w", encoding="utf-8")
```

```python
frases = ["Hola mundo python\n",
          "MI no bre es lyffy y aprendo python\n",
          "quiero ser el mejor programador"]

for frase in frases:
    archivo.write(frase)

archivo.close()
```

## Escribiendo varias frases con la funcion open() y writelines()

```python
archivo = open("mi_archivo.txt", mode="w", encoding="utf-8")

frases = ["Hola mundo python\n",
          "MI no bre es lyffy y aprendo python\n",
          "quiero ser el mejor programador"]

```

```python
archivo.writelines(frases)
archivo.close()
```

## Adicionando texto a un archivo ya existente

```python
archivo = open("mi_archivo.txt", mode="a", encoding="utf-8")
```

```python
archivo.write("\n \n Estoy aprendiedno a manipular archivos txt")
archivo.close()
```

## Creando / escribiendo en archivos con `with`

```python
with open("mi_archivo.txt", mode="a", encoding="utf-8") as f:
    f.write("\n Escribiendo un archivo txt con with")
```
