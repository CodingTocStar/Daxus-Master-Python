# Funciones lambda

## Definicion
* Funciones anonimas pequeñas que no requieren nombre
* Se pueden escribir en una linea
* Rapidas y cortas que generalmente se pasan como argumentos a otras funciones.

## Sintaxis



lambda argumentos: expresion

* argumentos: parametros de la definicion de la funcion
* expresion: donde se evalua y se devuelve la expresion

## Compracion entre funciones def y lambda



## Funcion lambda

```python
def sumar(a,b):
    return a + b

print(sumar(5,18))
print(type(sumar))
```

```python
# Funcion Lambda

type(lambda a, b : a + b)
```

```python
sumar_lambda = lambda a, b : a + b

sumar_lambda(5,18)
```

```python
multiplicar = lambda x,y: x * y

multiplicar(4,3)
```

```python
# Sumar con args
sumar_lambda_2 = lambda *args: sum(args)

print(sumar_lambda_2(1,2,3))
print(sumar_lambda_2(1,2,3,56,70,234))
```
