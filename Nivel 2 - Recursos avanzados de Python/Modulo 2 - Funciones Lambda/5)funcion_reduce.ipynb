# funcion reduce()

* Es parte del modulo `functool`
* Se utiliza para aplicar una funcion de forma acumulativa a los elementos de un iterable
* Asi se va reduciendo el iterable a un solo valor



* funcion: 
    * Toma dos argumentos
    * Devuelve un valor.
    * Se aplica de manera acumulativa a los elementos del iterable.
    * No quiere decir que tenga 2 elementos el iterable, sino que lo sigue haciendo
* iterable: Pueden ser 
    * Lista
    * tupla
    * conjunto
    * Sus elementos seran procesador por la funcion
* valor_inicial: (**opcional**)
    * Se proporciona como una semilla
    * si se proporciona --> se usa cvomo el primer valor en la acumulacion

```python
from functools import reduce
```

```python
lista = [i for i in range(1,6)]
# va tomando de a dos valores
reduce(lambda x,y: x+y, lista)
```

```python
reduce(lambda x, y: x*y, lista)
```

```python
# Usando reduce con un valor inicial

numeros = [i for i in range(1,6)]
resultado = reduce(lambda x,y: x+y, numeros, 10)

print(resultado)
```

```python
lista = [chr(letra) for letra in range(97,103)]
print(lista)

reduce(lambda x,y: x+y, lista)
```
