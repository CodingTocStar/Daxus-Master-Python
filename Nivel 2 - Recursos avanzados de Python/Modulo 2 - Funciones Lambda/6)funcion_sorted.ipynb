# Funcion sorted()

* Sirve para ordenar cualquier iterable de manera **no destructiva** 
* Devuelve una **nueva lista ordenada**
* El iterable original permanece **sin cambios**

## Sintaxis



* iterable: El que queres ordenar
* key(**opcional**): 
    * Una funcion que toma un solo argumento
    * Devuelve un valor para ser usado como **clave de comparacion**
    * Los elementos del iterable se ordenan en funcion de este valor.
* reverse (**opcional**): 
    * Si se estable True, la lista se ordena en **orden ascendente**
    * El valor predeterminado es False(orden ascendente)

```python
nombres = ["Juan", "Alex", "zai", "sebastian", "jhoan", "Jarson"]
nombres
```

```python
print(sorted(nombres))
print(sorted(nombres, reverse=True))
```

```python
nombres_edades = [("Juan", 25), ("Alex", 30), ("zai", 28), ("sebastian", 24), ("jhoan", 35), ("Jarson", 56)]
nombres_edades
```

```python
print(sorted(nombres_edades, key=lambda persona: persona[1])) # accede al elemento 1 de la tupla, o sea la edad
# compara segun la edad esta key

print(sorted(nombres_edades, key=lambda persona: persona[1], reverse=True))
```

```python
"""
Recordar que ele metodo sort() de las listas, modifica el elemento original
"""

nombres = ["Juan", "Alex", "zai", "sebastian", "jhoan", "Jarson"]
print(nombres)

nombres.sort()

print(nombres)
```
