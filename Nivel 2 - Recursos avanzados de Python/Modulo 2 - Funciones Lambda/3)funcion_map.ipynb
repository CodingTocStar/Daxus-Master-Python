# Funcion map()

* Aplica una funcion especifica a <u>todos los elementos de un iterable</u>
* Devuelve un objeto map, que es un **iterador**

## Sintaxis

map(funcion, iterable)



* funcion: la que se aplica a cada elemento del iterable
    Puede ser:
    * built-in
    * Definida por el usuario
    * funcion lambda
* iterable: pueden ser
    * lista
    * tupla
    * conjunto
