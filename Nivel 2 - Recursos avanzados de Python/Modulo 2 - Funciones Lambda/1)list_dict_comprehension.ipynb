# List comprehension y Dict comprehension

* List comprehension
    * Forma concisa y elegante de crear listas.
    * Permite generar una nueva lista aplicando una expresion a cada elemento de una secuencia o iterable en una sola linea.



### Sintaxis condicional





```python
#crear lista sin list comprehension

lista = []

for i in range(1,11):
    lista.append(i)
    
print(lista)
```

```python
#Creando lista con list comprehension

lista_numeros = [i for i in range(1,11)]

print(lista_numeros)
```

```python
lista_numeros = [i for i in range(1, 11)]

numeros_cuadrados = [elemento **2 for elemento in lista_numeros]

print(numeros_cuadrados)
```

```python
print(lista_numeros)
```

```python
numeros_pares = [elemento for elemento in lista_numeros if elemento % 2 == 0]

print(numeros_pares)
```

## dict comprehension

* Es una forma concisa y elegante de crear diccionarios
* Usa pares clave:valor



### Sintaxis condicional



{clave: valor if condicion else not_condicion for elemento in iterable}

```python
dic_original = {"a": 1, "b":2, "c":3}

dic_invertido = {v:k for k, v in dic_original.items()}

print(dic_invertido)
```

```python
frutas = ["Manzana", "banano", "cereza"]

longitudes = {fruta:len(fruta) for fruta in frutas}

print(longitudes)
```

```python
clasificacion = {x:"par" if x %2 == 0 else "impar" for x in range(0,5)}
print(clasificacion)
```
