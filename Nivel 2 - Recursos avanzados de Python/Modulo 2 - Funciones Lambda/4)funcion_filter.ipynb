# FUncion filter

* Se usa para filtrar elementos de un iterable en funcion de una condicion especificada.
* Devuelve un nuevo iterable (tipo filter)
* Contiene los elementos para los cuales la funcion dada devuelve True

## Sintaxis



filter(funcion, iterable)

* funcion:
    * Recibe **un solo elemento** del iterable
    * Devuelve un valor True o False, que indica si el elemento debe ser incluido en el resultado
* Iterable: Pueden ser
    * lista
    * tupla
    * conjunto

Cuyos elementos seran evaluados por la funcion

```python
lista = list(range(1,101))

print(lista)
```

```python
lista_pares = list(filter(lambda num: num % 2 == 0, lista))

print(lista_pares)
```

```python
notas = [2,3, 3.4, 1.1, 2.8, 4.5, 4.3,2.3, 3]
notas

```

```python
from statistics import mean

media = mean(notas)
print(media)
```

```python
# filtrar las notas por encima del promedio

mejores_notas = list(filter(lambda nota: nota > media, notas))

print(f"Media: {media}")
print(f"Notas por encima de la media: {" - ".join([str(i) for i in mejores_notas])}")
```

```python
#  FIltraer las notas pro debajo de la media

notas_debajo_media = list(filter(lambda nota: nota <= media, notas))

print(f"Media: {media}")
print(f"Notas por debajo de la media: {" - ".join([str(i) for i in notas_debajo_media])}")
```
