# Manipulacion de fechas con `strptime()` y `strftime()`

## COnvirtiendo `str` a `datetime` con `strptime()`

```python
from datetime import datetime
```

```python
datetime_str = "31/01/25 23:59:28"

print(datetime.strptime(datetime_str, "%d/%m/%y %H:%M:%S"))
print(type(datetime.strptime(datetime_str, "%d/%m/%y %H:%M:%S")))
print(datetime.strptime(datetime_str, "%d/%m/%y %H:%M:%S").date())
```

```python
fecha_datetime = datetime.strptime(datetime_str, "%d/%m/%y %H:%M:%S")

print(type(fecha_datetime))
print(fecha_datetime)
print(fecha_datetime.year)
print(fecha_datetime.month)
print(fecha_datetime.day)
print(fecha_datetime.hour)
print(fecha_datetime.minute)
print(fecha_datetime.second)
```

```python
#Los formatos tienen que coincidir o salta erro

datetime_str = "31-01-25 23:59:28"

print(datetime.strptime(datetime_str, "%d/%m/%y %H:%M:%S"))
```

## COnvirtiendo `datetime` a `str` con strftime()

```python
fecha = datetime(2025, 8, 25, 16, 36, 15)

print(f"La fecha y la hora son: {fecha.strftime("%A %d %B %Y. %I:%M%p")}") #la p indica si es pm o am
```

```python
import locale
locale.setlocale(locale.LC_ALL, "es-Es.UTF-8")
```

```python
print(f"La fecha y la hora son: {fecha.strftime("%A %d %B %Y. %I:%M%p")}")  # la p indica si es pm o am
```

```python
locale.setlocale(locale.LC_ALL, "en-US.UTF-8")
```

```python
# la p indica si es pm o am
print(f"La fecha y la hora son: {fecha.strftime("%A %d %B %Y. %I:%M%p")}")
```

## COnvirtiendo `str` a `datetime` con `strptime()` y luego convirtiendolo `datetime` a `str` con `strftime()`

### Ejemplo
* FUente de datos: Base de datos
* Procesamiento: Python
* Destino: Dashboard

```python
datetime_str = "31-01-22 23:59:28"
```

```python
datetime_str_py = datetime.strptime(datetime_str, "%d-%m-%y %H:%M:%S")

print(datetime_str_py)
print(type(datetime_str_py))
```

```python
datetime_str_dash = datetime_str_py.strftime("%A %d %B %Y, %I:%M%p")
print(datetime_str_dash)
```
