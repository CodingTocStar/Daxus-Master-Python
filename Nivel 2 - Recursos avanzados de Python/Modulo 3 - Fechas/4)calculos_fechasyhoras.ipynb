# Realizando calculos con fechas y horas

```python
import pytz
from datetime import date, datetime, timedelta
```

```python
fecha_inicial = date(2025,1,1)
fecha_final = date(2025,3,15)

diferencia_dias = fecha_final - fecha_inicial

print(diferencia_dias)
print(type(diferencia_dias))
print()
print(diferencia_dias.days)
print(type(diferencia_dias.days))
```

```python
suma_1dia = timedelta(days=1)
print(type(suma_1dia))
```

```python
datetime.now().date() + suma_1dia
```

```python
#Sumar 3 horas a un datetime

datetime(2025, 6,13, 11, 34, 50) + timedelta(hours=3)
```

```python
# Sumar 4 dias y 3 horas a un datetime

datetime(2025, 6, 13, 11, 34, 50) + timedelta(hours=3, days=4)
```

```python
fecha_inicial = datetime(2025, 4, 4 ,21, 50, 10)
fecha_final = datetime(2025, 4, 5 ,19, 30, 28)
```

```python
diferencia = fecha_final - fecha_inicial

print(diferencia)
print(diferencia.days)
#print(diferencia.hours) No tiene ese atributo
print(diferencia.seconds)
```
