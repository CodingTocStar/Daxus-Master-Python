## Clase `date`

```python
#import datetime

from datetime import date #para no tener que traer las dos y solo usar date
```

```python
date(year=2025, month=8, day=25)
```

```python
type(date(year=2025, month=8, day=25))
```

```python
print(date(2025, 8, 25))
```

## Creando fechas

```python
fecha = date(year=2025, month=1,day=1)

#print(type(fecha))
#print(fecha)

print(f"Fecha Completa: {fecha}")
print(f"Año de la fecha: {fecha.year}")
print(f"Mes de la fecha: {fecha.month}")
print(f"Dia de la fecha: {fecha.day}")
print(type(fecha.year))
```

## MEtodo `today`

```python
fecha_actual = date.today()

print(fecha_actual)
print(type(fecha_actual))
```

## Metodo `weekday`



```python
date.weekday(fecha_actual)
```
