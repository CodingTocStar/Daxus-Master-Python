# Clase `datetime`

```python
from datetime import datetime
```

```python
datetime(year=2025, month=8, day=25, hour=14, minute=23, second=56, microsecond=0)
```

```python
date_time = datetime(2025, 8,25, 14, 23,56, 0)

print(f"datetime completo: {date_time}")
print(f"AÑO datetime completo: {date_time.year}")
print(f"Mes datetime completo: {date_time.month}")
print(f"Dia datetime completo: {date_time.day}")
print(f"HORA datetime completo: {date_time.hour}")
print(f"MINUTOS datetime completo: {date_time.minute}")
print(f"SEGUNDOS datetime completo: {date_time.second}")
```

## Metodo `date`

## Metodo `today`

```python
date_time.today()
```

## Metodo `weekday`

```python
datetime.weekday(date_time)
```

## Metodo time

```python
date_time.time()
```

## Metodo `now`

```python
print(datetime.now())
print(datetime.now().time())
print(datetime.today().time())
```
