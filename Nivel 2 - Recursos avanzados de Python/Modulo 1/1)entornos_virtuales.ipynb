# Entornos virtuales (VE) en python

* entornos aislados para gestionar dependencias y librerias de manera independiente para cada proyecto
* SI las instalas globalmente pueden generar conflictos cuando tengas dos proyectos con librerias de veriones distintas.

## Ventajas de utilizar VE

* Aislamiento de proyectos con diferentes versiones y evitar conflictos
* Asegura la reproducibilidad de los proyectos para todos los colaboradores.
* Mantiene la limpieza del sistema al no acumular libnrerias instaladas globalmente que generen conflictos.
