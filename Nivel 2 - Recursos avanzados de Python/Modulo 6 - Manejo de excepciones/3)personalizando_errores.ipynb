# Personalizando excepciones con `raise`

* La palabra clave `raise` se usa para lanzar una excepcion en el flujo de ejecucion del programa.
* Se puede usar para:
    * Volver a lanzar una **excepcion**
    * Lanzar una excepcion **personalizada**  

```python
def division(a,b):
    if b == 0:
        raise Exception("El divisor no puede ser igual a 0")
        #raise ZeroDivisionError("Recuerda no dividir por 0, gil")
    return a/b
    
```

```python
division(12,0)
```

```python
def dividir_enteros(a: int, b:int):
    try:
        if b == 0:
            raise ZeroDivisionError("No dividas por cero te dije!\n Es una operacion indeterminada... \n")
        if not isinstance(a, int) or not isinstance(b, int):
            raise ValueError("Los valores deben ser enteros, capo. \n")
    except Exception as e:
        print(f"Ocurrio un errro. Error: {e}")
    else:
        resultado = a/b
        print(f"El resultrado de la division '{a}/{b}' es: {resultado}")
    finally:
        print("Saliendo del programa")
```

```python
dividir_enteros(23,0)
```

```python
dividir_enteros(23.4, 4)
```

```python
dividir_enteros(45, 4)
```
