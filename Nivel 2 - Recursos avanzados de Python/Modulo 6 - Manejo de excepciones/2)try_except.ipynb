# Tratamiento de errores con `try...except`

## Sintaxis

try:
    coidgo a intentar ejecutar
except:
    Codigo ejecuytado al capturar la excepcion
else:
    codigo ejecutado si no se captura ninguna excepcioin (**opcional**)
finally:
    bloque ejecutado siempore (**opcional**)

## Else y finally

```python
try:
    resultado = 5/2
except ZeroDivisionError:
    print("NO es posible esta diviision por cero (indeterminado)")
else:
    print(f"El resuultado de la division es: {resultado}")
finally:
    print("Saliendo del programa")
```

```python
def division_segura(a,b):
    try:
        resultado = a/b
    except ZeroDivisionError as e:
        print(f"No es posible desarrollar esta divisiion (indeterminado). Error: {e}")
    except ValueError as e:
        print(f"Valor(es) ingresados no son valido(s). Error {e}")
    except Exception as e:
        print(f"Ocurrio un error. Error {e}")
    else:
        print(f"El resuultado de la division es: {resultado}")
    finally:
        print("Saliendo del programa")
```

```python
division_segura(23.5, 2)
print()

division_segura(10,0)
print()

division_segura("juancho", 2)
print()

print("El programa continua")
```
