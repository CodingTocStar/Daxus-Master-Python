# Manejo avanzado de excepciones con la libreria `logging`

```python
import logging
```

```python
"""
Niveles de loggin
- Debug
- Info
- Warning
- Error 
- Critical
"""

#Handler que guarda la configuracion y va a guardar los evento en un archivo

logging.basicConfig(level=logging.DEBUG,
                    format="%(asctime)s - %(levelname)s - %(message)s",
                    filename="eventos.logging",
                    filemode="a")
```

```python
# Crear un nuevo Handler y sus caracteristicas para la consola
console_handler = logging.StreamHandler()
console_handler.setLevel(level=logging.DEBUG)

console_formatter = logging.Formatter("%(asctime)s - %(levelname)s - %(message)s")


logging.getLogger().addHandler(console_handler)
```

```python
a = 45
b = 2

try:
    if b == 0:
        logging.error("Division por cero")
        raise ZeroDivisionError("No olvides que no se puede dividir por cero! Es una operacion indeterminada...\n")
    if not isinstance(a, int) or not isinstance(b, int):
        logging.debug("No se ingresaron valores enteros.")
        raise ValueError("Los valores deben ser enteros...")
except Exception as e:
    logging.error(f"{e}")
    print(f"Ocurrio un error, {e}")
else:
    resultado = a/b
    logging.info(f"La division entre '{a}/{b}' es igual a {resultado}")
    print(f"La division entre '{a}/{b}' es igual a {resultado}")
finally:
    print("Saliendo del programa")
```
