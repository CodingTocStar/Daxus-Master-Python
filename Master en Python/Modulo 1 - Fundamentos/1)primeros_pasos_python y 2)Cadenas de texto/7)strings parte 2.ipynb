# **Metodos** para *manipular* str

## split()

* divide un str convirtiendolo  en lista
* opcional podes indicar el caracter con el que vas a hacer la separacion

```python
texto = "hola mundo Python"

print(texto.split())
```

```python
frutas = "fresa, mango, maracuya, sandia"

print(frutas.split())

print(type(frutas.split(sep=",")))
```

## strip()

Elimina espacios en blanco o caracteres especificos al principio y al final de la cadena

```python
texto= "              hola mundo    "

print(texto)
print(texto.strip())
```

```python
texto ="////// bienvenidos al curso de python/////"

print(texto)
print(texto.strip("/"))
```

## lstrip()

Elimina espacios o caracteres al principio de la cadena (l de left)

```python
print(texto.lstrip("/"))
```

## rstrip()

lo mismo pero a la derecha


```python
print(texto)
print(texto.rstrip("/"))
```

## capitalize()

Pone el primer caracter de la cadena en mayuscula

```python
nombre = "goku"

print(nombre.capitalize())
```

## lower()

pone todos los caracteres en minuscula

```python
nombre = "GOKU"

print(nombre.lower())
```

## upper()

```python
texto = "bienvenidos al curso"

print(texto.upper())
```

## title()

```python
nombre_completo = "son goku"

print(nombre_completo.title())
```

## replace()

remplaza todas las ocurrencias de una subcadena vieja por otra subcadena nueva

```python
texto ="Hola mundo"

print(texto.replace("mundo", "Python"))
```
