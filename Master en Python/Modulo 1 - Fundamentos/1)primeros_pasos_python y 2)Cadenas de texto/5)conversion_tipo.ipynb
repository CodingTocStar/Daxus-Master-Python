Conversion dde tipos o casting

casting a int

```python
numero_float = 3.14

print(int(numero_float))
```

```python
numero_str = input("ingrese un numero: ")

print(int(numero_str))

print(type(int(numero_str)))
```

```python
booleano = False
print(int(booleano))
```

```python
string = "Shimbo"

print(int(string)) #no se puede
```

Casting a float

```python
num_int = 5

print(float(num_int))
```

```python
num_str = "3.1416"

print(float(num_str))
```

```python
booleano = True
print(float(booleano))
```

```python
string = "python"

print(float(string))
```

Casting a bool


```python
num_int = 0

print(bool(num_int))
```

```python
numero_float = 1.1

print(bool(numero_float))
```

```python
string_vacio = ""

print(bool(string_vacio))
```

```python
string = "hola"

print(bool(string))
```

```python
num_float = 2.26

print(str(num_float))

print(type(str(num_float)))
```
