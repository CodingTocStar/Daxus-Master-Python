Strings


Se usa principalmente para guardar texto.
Esta conformado por una serie de caracteres.



str son case sensitive

```python
"Juancho" == "juancho"
```

Concatenar str con +

```python
nombre = "Son"
apellido = "Goku"

print(nombre, apellido)
```

```python
print(nombre + " " + apellido)
```

```python
type("a")
```

```python
print("a" + "b"+ "c" + "d")
```

str son objetos iterables

```python
print(apellido)
print(apellido[0])
print(apellido[1])
```

```python
lenguaje = "Mi nombre es Juancho H y estoy estudiando Python"
print(lenguaje[-6])
print(lenguaje[-5])
print(lenguaje[-4])
print(lenguaje[-3])
print(lenguaje[-2])
print(lenguaje[-1])
```
