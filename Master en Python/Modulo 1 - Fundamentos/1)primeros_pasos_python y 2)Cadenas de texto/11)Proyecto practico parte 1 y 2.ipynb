# Sistema de control de inventario

## Funcionalidades

* Mostrar mensaje de nienvenida uasndo print()
* Pedir al usuario que ingrese el nombre del producto, la cantaidad y el precio unitario usando input()
* Almacenar la informacion en variables
* Realizar el calculo del valor total del inventario inicial (cantidad x precio)

```python
print("Bienvenido al Sistema de Control de Inventario")
nombre_producto = input("Ingrese el nombre del producto: ")
cantidad_producto = int(input("Ingrese la cantidad: "))
precio_unitario = float(input("Ingrese precio unitario: "))

print("-" * 20)

valor_total = precio_unitario * cantidad_producto

print(f"Producto       : {nombre_producto.strip().title()}")
print(f"Cantidad       : {cantidad_producto}")
print(f"Precio unitario: ${precio_unitario:.2f}")
print("-" * 10)
print(f"Total: ${valor_total:.2f}")
```
