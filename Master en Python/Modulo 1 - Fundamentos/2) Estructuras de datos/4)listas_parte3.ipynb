## Concatenacion (+) y repeticion (*) de listas

```python
lista_1 = [1,2,3,4]
lista_2 = [5,6,7,8]

lista = lista_1 + lista_2

print(lista)
```

```python
lista = ["Homero", "Marge", "Lisa", "Bart"]
lista_repeticion = lista * 3

print(lista_repeticion)
```

## Metodo len()

Indica el numero de elementos que contiene un iterable (ahora list)

```python
print(lista)
print(len(lista))
```

```python
print(lista_repeticion)
print(len(lista_repeticion))
```

## Metodo index()

* Se utiliza para encontrar el indice de la **primera aparicion** de un elemento de la lista.
* Si el elemento no esta en la lista, pytho  tira un error lammado **ValueError** 

```python
print(lista)

print(lista.index("Bart"))
print(lista.index("Lisa"))
```

```python
print(lista.index("Barnie"))
```

## Metodo sorted()

* sorted **NO es un metodo de listas**
* Es una funcion **built-in**
    * Devuelve una nueva lista ordenada
    * DEeja la lista original sin cambios

```python
lista_desordenada = [1,5,3,2,4]
nueva_lista_ordenada = sorted(lista_desordenada)

print(lista_desordenada)
print(nueva_lista_ordenada)
```

## Metodo alternativo con indices para `reverse()`

```python
lista = [1,2,3,4,5,6]

print(lista[::-1])
```

## Metodo `copy()`

* Crea una copia superficial (*shallow copy*) de una lista.
* El nuevo objeto copiado es una lista, pero sus elementos no son copiados profundamente
    * Si son <u>Objetos mutables</u> (otras listas) -> se copian como referencias
    * Si son <u>objetos inmutables</u> (numeros o cadenas) -> se copian como valores
