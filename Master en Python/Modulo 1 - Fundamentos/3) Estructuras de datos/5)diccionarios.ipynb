# Diccionarios `dict{key : value}`

* Almacenan datos en pares **clave-valor**.
* Muy versatiles debido a que se pueden acceder a los datos rapidamente.

* A diferencia de las listas que usan indices numericos, aca usamos claves.
    * Las claves son *inmutables* y de **cualquier tipo** (cadenas, numero o tuplas).

## Caracteristicas

* **Desordenados**: Se van agregando a medida que se crean.
* **Acceso rapido**: Acceso rapido debido a que internamente usan estructura de datos llamada tabla hash para almacenar las claves y sus valores asociados.
* **Claves unicas**: Deben ser unicas. Si asignas un valor a una clave existente se sobrescribe.
* **Valores mutables**: Pueden ser de cualquier tipo.




## Creando un `dict`

```python
mi_dict = {
    "Nombre": "Goku",
    "Apellido": "Son",
    "Raza": "Sayajin",
    "Planeta": "Vegita",
    "Edad": 25
}

mi_dict
```

```python
type(mi_dict)
```

## Accediendo a claves, valores e items

```python
mi_dict = {
    "Nombre": "Goku",
    "Apellido": "Son",
    "Raza": "Sayajin",
    "Planeta": "Vegita",
    "Edad": 25,
}

mi_dict
```

## Metodo keys()

Devuelve una vista de todas las claves del diccionario

```python
print(mi_dict.keys())
print(type(mi_dict))
```

## Metodo `values()`

Devuelve una vista con todos los valores

```python
mi_dict.values()
```

## Metodo `items()`

Devuelve una vista de todos los pares clave-valor

```python
mi_dict.items()
```

# Accediendo a los elementos de un `dict`


## LLamada de la `clave/llave`

```python
datos_personales = {
    "Nombre": "Goku",
    "Apellido": "Son",
    "Raza": "Sayajin",
    "Planeta": "Vegita",
    "Edad": 25,
}

datos_personales
```

```python
print(datos_personales["Apellido"])
print(datos_personales["Planeta"])
```

## Metodo `get()`

* Devuelve el valor de una clave especifica
* Si la clave **NO existe**
    * Devuelve `None` o un valor predeterminado que yo especifiqie

```python
print(f"Planeta: {datos_personales.get("Planeta")}")
print(f"Nombre: {datos_personales.get("Nombre")}")
print(f"Edad: {datos_personales.get("Edad")}")
```

## INsertando elementos en un `dict`

```python
datos_personales
```

```python
datos_personales["Hijos"] = ["Gohan", "Goten"]
```

```python
print(datos_personales["Hijos"][0])
print(datos_personales["Hijos"][1])
```

## Actualizando elementos en un `dict` 

```python
datos_personales["Edad"] = 26
datos_personales["Nombre"] = "Goku Rafael"
```

## Eliminando elementos de un `dict`

### Metodo `pop()`
* Elimina y devuelve el valor asociado a la clave dada.
    * Si la clave **NO existe** Y no se proporciona un valor default =>
        * Lanza una excepcion/error `KeyError`

```python
datos_personales.pop("Planeta")
```

```python
datos_personales
```

### Metodo `popitem()`

Elimina y devuelve el ultimo par clave-valor del diccionario (Antes eliminaba un par aleatorio)

```python
hijos = datos_personales.popitem()

print(hijos)

datos_personales
```

### Metodo `clear()`
Elimina todos los elementos del diccionario

```python
datos_personales.clear()
datos_personales
```
