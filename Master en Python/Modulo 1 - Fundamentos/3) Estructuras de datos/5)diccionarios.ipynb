# Diccionarios `dict{key : value}`

* Almacenan datos en pares **clave-valor**.
* Muy versatiles debido a que se pueden acceder a los datos rapidamente.

* A diferencia de las listas que usan indices numericos, aca usamos claves.
    * Las claves son *inmutables* y de **cualquier tipo** (cadenas, numero o tuplas).

## Caracteristicas

* **Desordenados**: Se van agregando a medida que se crean.
* **Acceso rapido**: Acceso rapido debido a que internamente usan estructura de datos llamada tabla hash para almacenar las claves y sus valores asociados.
* **Claves unicas**: Deben ser unicas. Si asignas un valor a una clave existente se sobrescribe.
* **Valores mutables**: Pueden ser de cualquier tipo.
