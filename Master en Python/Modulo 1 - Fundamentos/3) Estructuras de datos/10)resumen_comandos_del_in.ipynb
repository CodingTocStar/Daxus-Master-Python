# Comandos `in` y `del`

## Aplica `in`

* Metodo directo para validar la existencia de un elemento en estructuras de datos.
* Especial en `sets` y `dicts` porque usan *tablas hash* (claves y valores)

* list
* dict -> validacion keys y values
* tuple
* set

## Aplica `del`

Util para prescindir de varialbes que almacenan grandes cantidades de datos cuando ya no son necesarias.

* list : eliminacion por *index* y lista completa
* dict: eliminacion por *key* y diccionario completo
* tuple: eliminacion completa unicamente (son ***inmutables***)
* set: eliminacion completa unicamente (son **desordenados**)

## List

### in

```python
personajes =  ["Naruto", "Goku", "Pikachu", "Luffy", "Bart"]


```

```python
"Naruto" in personajes
```

```python
"Vegeta" in personajes
```

### `del`

```python
print(personajes)
```

```python
del personajes[-2]
```

```python
print(personajes)
```

```python
del personajes

print(personajes)
```

## Dict

### in

```python
datos_personajes = {
    "Nombre": "Goku",
    "Apellido": "Son",
    "Edad": 25,
    "Raza": "Sayajin",
    "Planeta": "Vegita"
}
```

```python
"Edad" in datos_personajes.keys() #Funciona keys y values
```

```python
"Vegita" in datos_personajes.values()
```

### `del`

```python
del datos_personajes["Planeta"]
print(datos_personajes)
```

```python
del datos_personajes
print(datos_personajes)
```
