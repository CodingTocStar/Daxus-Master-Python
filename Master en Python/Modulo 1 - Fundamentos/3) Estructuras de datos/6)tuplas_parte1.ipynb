# Tuplas - `tuple()`

* Parecidas a las listas pero son **INMUTABLES**
* Muy utiles cuando hay que asegurarse que los datos:
    * **NO** sean modificados
    * Mayor eficiencia en terminos de memoria

## Principales caracteristicas

* Definidas con ()
* <u>Inmutables<u> --> seguridad (no se pueden agregar ni quitar elementos)
* **ordenadas** --> podes acceder a los elementos por su indice.
* Pueden contener **cualquier tipo de dato (numeros, cadenas, listas, etc)**
* Al ser inmutables, las tuplas son mas rapidas que las listas para acceso.

## Creando una tupla - `tuple`

```python
usuario= ("Juancho", "Hernandez", 25, False)

print(usuario)
print(type(usuario))
```

## Accediendo a los elementos de una `tuple`

```python
print(usuario[0])
print(usuario[1])
print(usuario[-1])
```

## Metodo `count()`

Devuelve el numero de veces que el valor X aparece en la tupla

```python
numeros = (1,2,3,4,5,6,7,6,6,7)

numeros.count(6)
```

## Metodo `index`

* Devuelve el indice del primer elemento cuyo valor sea x.
* Si el valor no se encuentra --> `ValueError`
* Se puede especificar un rango de busqueda usando `start` y `end`

```python
print(numeros)

numeros.index(7)
```

## Las tuplas son **inmutables!!**

```python
print(usuario)
```

```python
usuario[0] = "Bimbo"
```

## Eliminacion

```python
del usuario[-1]
```
