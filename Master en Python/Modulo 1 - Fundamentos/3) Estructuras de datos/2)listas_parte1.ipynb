# Listas - list[]

* Es una coleccion *ordenada* y **mutable** de elementos 
* Muy flexible
* Se pueden almacenar distintos tipos de datos: 
    * enteros
    * cadenas de texto
    * mas listas
    * y asi

## Principales caracteristicas

* Ordenadas -> los elementos tienen un *indice*
* mutables -> se puede modificar el contenido
* permiten duplicados -> pueden haber elementos repetidos

# Creando lista

```python
lista_dbz = ["Goku", "Gohan", "Vegeta"]

print(type(lista_dbz))
print(lista_dbz)
```

# Lista con distintos tipos de datos

```python
lista_heterogenea = [25, 25,  True, "Hola", 1.80, "Python"]

print(type(lista_heterogenea))
print(lista_heterogenea)
```

```python
lista_anidada = [2000, "juancho", [1,23,24, True]]

print(type(lista_anidada))
print(lista_anidada)
```

# Accediendo a un elemetno de la lista por medio del indice

```python
lista_dbz = ["Goku", "Gohan", "Vegeta", "Trunks"]

print(lista_dbz[0])
print(lista_dbz[1])
print(lista_dbz[2])
print(lista_dbz[-1])
```

```python
print(lista_anidada[1])
print(lista_anidada[2])
print(lista_anidada[2][-1])
```

# Utilizando slicing

```python
personajes = ["Naruto", "Luffy", "Pikachu", "Bart", "Homero", "Usopp"]

print(personajes)
print("------------------")
print(personajes[0:2])
print(personajes[-3:])
print(personajes[2:4]) # es excluyente el extremo superior
```
