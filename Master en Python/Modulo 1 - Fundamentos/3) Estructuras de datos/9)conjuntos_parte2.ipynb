# Transformando un set a `list`

```python
conjunto = {1,2,3,4,5}

print(conjunto)
print(type(conjunto))
```

```python
conjunto_a_lista = list(conjunto)

print(conjunto_a_lista)
print(type(conjunto_a_lista))
```

## Metodo `discard()`

* Elimina el elemento especificado del conjutno.
* Si no esta presente --> no hace nada y tira error

```python
conjunto = {1,2,3, False, "Python"}
print(conjunto)
```

```python
conjunto.discard(3)
print(conjunto)
```

```python
conjunto.discard("Java")
print(conjunto)
```

```python
conjunto.discard(False)
print(conjunto)
```

## Metodo `difference()`

* Devuelve un nuevo conjunto con los elementos que estan en el original, **pero NO** en los otros conjuntos que se pasen como argumento.
* No modifica el conjunto original

```python
conjunto_1 = {1,2,3,4}
conjunto_2 = {3,4,5,6}

diferencia = conjunto_1.difference(conjunto_2)
print(diferencia)
```

## Metodo `issubset()`

* Verifica si el conjunto actual es un subconjunto del conjunto proporcionado.
* Devuelve `True` si todos lso elementos del conjunto actual estan en el conjunto pasado como argumento.


```python
conjunto_1 = {1,2}
conjunto_2 = {1,2,3,4}
print(conjunto_1.issubset(conjunto_2))
```

```python
conjunto_3 = {5,6}
print(conjunto_3.issubset(conjunto_1))
```

## Metodo `issuperset()`

* Verifica si el conjunto actual es superconjunto del proporcionado.
* Devuelve true si **TODOS** los elementos del conjunto pasado como argumento estan en el conjunto original

```python
conjunto_1 = {1,2,3,4}
conjunto_2 = {1,2}

print(conjunto_1.issuperset(conjunto_2))
```

```python
conjunto_3 = {5,6}
print(conjunto_1.issuperset(conjunto_3))
```
