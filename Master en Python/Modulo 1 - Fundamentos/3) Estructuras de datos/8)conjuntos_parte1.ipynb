# COnjuntos - `set()`

* son una estructura d datos que almacena elementos **<u>unicos</u>**

* Sin orden especifico.
* Definidos por {} o la funcion `set()`
* **<u>NO permiten elementos duplicados</u>**

## Principales caracteristicas

* Son **no ordenados** --> **NO** se acceden por <u>indice</u>
* **Mutables** --> *PERO...* no permiten <u>duplicados</u>
* **Iterables** --> Se pueden <u>recorrer</u> estos objetos

## Creando un `set`

```python
conjunto = {1,2,3,4,5}

print(conjunto)
print(type(conjunto))
```

```python
conjunto_repetido = {1,2,2,3,4,5}

print(conjunto_repetido) ## TOma solo el primero
```

## Manipulando los elementos de un `set`

### MEtodo `add()` --> SE pueden agregar de distintos tipos!

* Agrega el elemento x al conjunto
* Si ya existe --> NO hace nada (nada de duplicados)

```python
print(conjunto)
```

```python
conjunto.add(6)

print(conjunto)
```

```python
conjunto.add("Guts")
print(conjunto)
```

```python
conjunto.add(False)
conjunto.add(True) #LO toma como 1 y no lo agrega
print(conjunto)
```

```python
conjunto.add(1.234)
print(conjunto)
```

```python
conjunto.add(("Naruto", "Uzumaki")) ## Podes meter tuplas

print(conjunto)
```

### NO permiten listas, las listas permiten duplicados

```python
# conjunto.add(["1,2,3"]) || TypeError unhasheable, ya estan presentes
# conjunto.add(["7,8,9"])
print(conjunto)
```
